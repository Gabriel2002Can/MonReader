```python
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim

import torchvision.transforms as transforms
from torchvision import datasets

print('Imported')
```

```python
transform = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,0.5,0.5), (0.5,0.5,0.5))
])
```

```python
train_data = datasets.ImageFolder(root='./images/training', transform=transform)

test_data = datasets.ImageFolder(root='./images/testing', transform=transform)
```

```python
type(train_data)
```

```python
print(train_data.classes)

print(test_data.classes)
```

```python
train_loader = torch.utils.data.DataLoader(train_data, batch_size=32, shuffle=True, num_workers=2)

test_loader = torch.utils.data.DataLoader(test_data, batch_size=32, shuffle=True, num_workers=2)
```

```python
`class NeuralNet(nn.Module):`
    `def __init__(self):`
        `super().__init__()`

        #           channels,feature maps, kernel size 
        self.conv1 = nn.Conv2d(3, 12, 3) # 128 - 3 = 125 + 1 = 126x126 = (12, 126, 126)
        self.pool = nn.MaxPool2d(2,2) # (12, 63, 63)
        self.conv2 =  nn.Conv2d(12, 24, 3) # 63 - 3 = 60 + 1 = 61x61 = (24, 61, 61) -> Pooling (24, 30, 30)
        # Pooling
        self.conv3 = nn.Conv2d(24, 48, 3) # 30 - 3 = 27 + 1 = 28x28 = (48, 28, 28) -> Pooling (48, 14, 14)
        # Pooling
        self.conv4 = nn.Conv2d(48, 96, 3) # 14 - 3 = 11 + 1 = 12x12 = (48, 12, 12) -> Pooling (48, 6, 6)
        # Pooling
        # Flatten
        self.fc1 = nn.Linear(96 * 6 * 6, 120)
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 2)

    def forward(self, X):
        X = self.pool(F.relu(self.conv1(X)))
        X = self.pool(F.relu(self.conv2(X)))
        X = self.pool(F.relu(self.conv3(X)))
        X = self.pool(F.relu(self.conv4(X)))
        X = torch.flatten(X, 1)
        X = F.relu(self.fc1(X))
        X = F.relu(self.fc2(X))
        X = self.fc3(X)
        return X
```

```python
net = NeuralNet()

loss_function = nn.CrossEntropyLoss()
optimizer = optim.SGD(net.parameters(), lr=0.001, momentum=0.9)
```

```python
for epoch in range(20):
    print(f'Training epoch {epoch}...')

    running_loss = 0.0

    for i, data in enumerate(train_loader):
        inputs, labels = data

        optimizer.zero_grad()

        outputs = net(inputs)

        loss = loss_function(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss =+ loss.item()

    print(f'Loss: {running_loss / len(train_loader):.4f}', end='\n\n')
```

```python
torch.save(net.state_dict(), 'trained_net.pth')
```

```python
net = NeuralNet()
net.load_state_dict(torch.load('trained_net.pth'))
```

```python
correct = 0
total = 0

net.eval()

with torch.no_grad():
    for data in train_loader:
        images, labels = data
        outputs = net(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100* correct/total

print(f'Accuracy Training: {accuracy}%')
```

```python
correct = 0
total = 0

net.eval()

with torch.no_grad():
    for data in test_loader:
        images, labels = data
        outputs = net(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100* correct/total

print(f'Accuracy Test: {accuracy}%')
```
